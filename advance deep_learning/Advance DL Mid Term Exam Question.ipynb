## Explain the concept of Convolutional Neural Networks (CNNs) in detail. Describe why CNNs are preferred over fully connected Artificial Neural Networks for image data, how local connectivity and weight sharing work, and how these ideas help in extracting spatial features from face images used for age-group detection.


# Write Answer 01:
A **Convolutional Neural Network (CNN)** is a type of deep neural network mainly used for **image processing and image classification**. It can automatically learn useful features from images.
CNN architecture in face detection.png
CNN is better than a fully connected ANN for image data because:
1. It preserves the **spatial information** of an image.
2. It uses **fewer parameters**.
3. It automatically learns image features.
4. It can detect the same feature at different positions.
### Local Connectivity
In CNN, each neuron looks at only a **small part of the image**, instead of the whole image.

A small filter, such as a **3×3 filter**, moves over the image and detects features like: **Edges → Lines → Curves** .
So, local connectivity helps CNN to learn **local spatial features**.
### Weight Sharing
CNN uses the **same filter weights** at different positions of an image.
For example, the same filter can detect an edge on the left, right, top, or bottom of a face.Therefore, weight sharing:
* reduces the number of parameters.
* helps detect the same feature at different locations.
### CNN for Age-Group Detection
For a face image, CNN extracts features step by step:
**Face Image → Edges/Curves → Eyes/Nose/Mouth → Face Shape/Skin Features → Age Group**

The early layers detect simple features such as edges and curves. Deeper layers detect complex facial features such as face shape, skin texture, and wrinkles. Finally, the classification layer predicts the age group, such as **Child, Young, Adult, or Old**.
### Simple CNN Architecture
**Input Face Image → Convolution + ReLU → Pooling → Convolution + ReLU → Pooling → Flatten → Fully Connected Layer → Age Group**




# Question 02: [ Marks 10 ]

## Describe the architecture of a basic CNN in detail. Explain the role of each component (convolutional layer, activation function, pooling layer, flatten layer, and fully connected layer). Also discuss how increasing the number of convolutional layers and filters affects feature extraction, model performance, and overfitting.


# Write Answer 02:
### Basic CNN Architecture
**Input Image → Convolution → ReLU → Pooling → Convolution → ReLU → Pooling → Flatten → Fully Connected → Output**

### 1. Convolutional Layer:
The convolutional layer is responsible for **feature extraction**. It uses small filters or kernels that move over the image and detect important features.

Early layers detect simple features such as **edges, lines, and curves**. Deeper layers can detect more complex features such as **eyes, nose, mouth, and shapes**.
### 2. Activation Function
An activation function adds **non-linearity** to the network. The most commonly used activation function in CNNs is **ReLU**.
**ReLU(x) = max(0, x)**
It changes negative values to zero and keeps positive values. This helps the CNN learn complex patterns.
### 3. Pooling Layer
The pooling layer reduces the size of feature maps while keeping important information.
A common method is **Max Pooling**, which selects the maximum value from a small region.
Pooling helps to:

* Reduce computation.
* Reduce the size of feature maps.
* Keep important features.
### 4. Flatten Layer
The flatten layer converts the 2D or 3D feature maps into a **one-dimensional vector**.
**Feature Maps → Flatten → 1D Vector**

This vector is then given to the fully connected layer.

### 5. Fully Connected Layer
The fully connected layer uses the extracted features to make the **final classification**.
For example, in face age-group detection, it can classify an image as:

**Child / Young / Adult / Old**

### Effect of Increasing Convolutional Layers and Filters
**Increasing convolutional layers:**
More layers allow the CNN to learn more complex and high-level features. However, too many layers can increase model complexity and may cause **overfitting**.

**Increasing filters:**
More filters allow a layer to learn more types of features. This can improve feature extraction and model performance. However, too many filters increase the number of parameters and computation, which can also increase the risk of **overfitting**.

## Explain what a filter (kernel) is in a CNN and how feature maps are generated during the convolution operation. Discuss how different filters can detect edges, textures, wrinkles, and other facial patterns that are useful for classifying a face as Young, Middle, or Old.


# Write Answer 03:
### Filter (Kernel) and Feature Maps in CNN

A **filter or kernel** is a small matrix of numbers used in a Convolutional Neural Network (CNN) to detect specific features from an image. Common filter sizes are 3×3 or 5×5.

During the **convolution operation**, the filter slides over different parts of the input image. The values of the filter and the corresponding image pixels are multiplied and added together. This produces new values that form a **feature map**.

The feature map shows where a particular feature is present in the image. A high value in the feature map means that the corresponding feature is strongly detected in that region.

For example:
Input Image-->Filter-->Convolution-->Feature Map

Different filters can detect different features. Some filters can detect **vertical and horizontal edges**, while others can detect **textures, lines, corners, wrinkles, and other facial patterns**.

In face classification, the early CNN layers usually detect simple features such as edges and textures. Deeper layers combine these simple features to learn more complex facial patterns such as eyes, nose, mouth, face shape, skin texture, and wrinkles.

For example, a filter may produce a strong feature-map response around wrinkle-like patterns. Other filters may detect facial edges and skin textures. The CNN combines these learned features and uses them to classify a face into categories such as **Young, Middle, or Old**.

### Example:

```text
Face Image
    ↓
Edge Filters → Detect edges
Texture Filters → Detect skin texture
Pattern Filters → Detect facial patterns
Wrinkle-related Filters → Detect wrinkle-like patterns
    ↓
Feature Maps
    ↓
CNN Classifier
    ↓
Young / Middle / Old
```

Thus, filters help CNNs extract useful features from images, and feature maps represent the locations and strength of those detected features.

## Explain padding and stride in CNNs in detail. Then explain the convolutional output size formula and what each symbol in the formula means. You do not need to derive the formula. Show how the output height and width change when you use valid padding, same padding, or a larger stride. Also explain why padding is useful when working with cropped face images.


# Write Answer 05:
**Padding** is the process of adding extra pixels around the border of an input image before performing convolution. Usually, zero values are added as padding.
Padding is useful because it helps preserve the spatial information of the image and prevents the output size from becoming too small too quickly. It also allows the filter to process features near the image boundaries.
There are two common types of padding:

**1. Valid Padding:**
In valid padding, no extra pixels are added.P=0

Therefore, the output size becomes smaller than the input size.

**2. Same Padding:**
Same padding adds enough pixels so that the output height and width remain the same as the input when the stride is 1.

### Stride

**Stride** is the number of pixels by which the filter moves across the input image during convolution.
Stride = 1: The filter moves one pixel at a time.
Stride = 2: The filter moves two pixels at a time.

### Convolutional Output Size Formula

The output size of a convolutional layer is calculated by:

$$
\boxed{
Output=
\left\lfloor
\frac{N+2P-F}{S}
\right\rfloor+1
}
$$

Where:

* **N** = input height or width
* **P** = padding
* **F** = filter or kernel size
* **S** = stride
* **Output** = output height or width
* **⌊ ⌋** = floor operation, which means taking the lower whole number.


### Example

Suppose an input image is 32×32 and the filter size is 3×3.

**Case 1: Valid Padding, Stride = 1**

Here, \(P=0\).Output=30

**Case 2: Same Padding, Stride = 1**

For a 3×3 filter, \(P=1\).

output =32

### Why Padding is Useful for Cropped Face Images

Padding is useful for cropped face images because important facial features may be located near the boundaries of the cropped image.
For example, **face edges, eyes, forehead, chin, wrinkles, and skin textures** may appear close to the image border. Without padding, these boundary features may be processed fewer times and can be lost more quickly as the feature maps become smaller.
Padding gives the convolutional filter extra space around the image. Therefore, it helps the CNN preserve and extract useful features from the entire cropped face.
In age detection, these features can help the CNN learn patterns related to different age groups such as **Young, Middle, and Old**.

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity when training an age-detection CNN on thousands of face images.


# Write Answer 06:
### Pooling in CNN

**Pooling** is a down-sampling technique used in Convolutional Neural Networks (CNNs). It reduces the spatial size of feature maps while keeping important information.
Pooling is usually applied after a convolutional layer. A small window, such as 2×2, moves over the feature map and converts several values into one value.
### MaxPooling

**MaxPooling** selects the maximum value from the pooling window.

For example:
```text
1  3
5  6
```
The maximum value is:Max=6

Therefore, MaxPooling keeps the strongest or most important activation in that region.

**Advantages of MaxPooling:**
* Preserves strong features.
* Reduces the size of feature maps.
* Reduces computation.
* Helps the CNN focus on important features.
* Provides some tolerance to small changes in feature position.
### AveragePooling

**AveragePooling** calculates the average of all values in the pooling window.

For example:

```text
1  3
5  6
```


Average==3.75
Thus, AveragePooling represents the overall or average information of a region.

**Advantages of AveragePooling:**

* Keeps the overall information of a region.
* Produces a smoother feature representation.
* Reduces the spatial size of feature maps.
* Can reduce computational requirements.

### How Pooling Reduces Computational Complexity

Suppose a feature map has size:

$$
64\times64=4096
$$

After 2×2 pooling with stride 2, the size becomes:

$$
32\times32=1024
$$

Therefore, the number of values is greatly reduced.
This means that the following CNN layers need fewer calculations and less memory. As a result, training becomes faster and more computationally efficient.

### How Pooling Helps Reduce Overfitting

Overfitting occurs when a CNN learns the training images too specifically and performs poorly on new images.Pooling reduces the spatial size of feature maps and removes some less important details. Therefore, the model becomes less dependent on exact pixel positions and small irrelevant variations.

For example, a small change in the position of an eye, wrinkle, or facial edge may not significantly change the pooled feature.
### Application to Age Detection
In an age-detection CNN trained on thousands of face images, pooling can reduce the size of feature maps after convolution.

For example:


Face Image
     ↓
Convolution
     ↓
Feature Maps
     ↓
MaxPooling / AveragePooling
     ↓
Smaller Feature Maps
     ↓
More CNN Layers
     ↓
Age Classification
     ↓
Young / Middle / Old


Therefore, pooling helps reduce computational complexity, memory usage, and overfitting, while keeping important facial features useful for age classification.

The Faces Age Detection dataset contains face images divided into three age classes: Young, Middle, and Old.

````python
import os
import random
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset # Added Dataset
from torchvision import datasets, transforms

from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
````

````python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("arashnic/faces-age-detection-dataset")

print("Path to dataset files:", path)
````

````python

data_dir = "/kaggle/input/faces-age-detection-dataset"
csv_path = os.path.join(data_dir, "faces", "train.csv")
image_dir = os.path.join(data_dir, "faces", "Train")

df = pd.read_csv(csv_path)

print("CSV shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
fig, axes = plt.subplots(2, 3, figsize=(10, 6))

for i, ax in enumerate(axes.flat):

    image_id = df.iloc[i]["ID"]
    label = df.iloc[i]["Class"]

    image_path = os.path.join(image_dir, str(image_id))

    image = Image.open(image_path).convert("RGB")

    ax.imshow(image)
    ax.set_title(label)
    ax.axis("off")

plt.tight_layout()
plt.show()
````

part B

````python
label_map = {
    "Young": 0,
    "Middle": 1,
    "Old": 2
}

df["label"] = df["Class"].str.title().map(label_map)

print(df.head())
````

````python
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])
````

````python
class AgeDataset(Dataset):

    def __init__(self, dataframe, image_dir, transform=None):
        self.df = dataframe
        self.image_dir = image_dir
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):

        image_id = str(self.df.iloc[index]["ID"])
        label = int(self.df.iloc[index]["label"])

        image_path = os.path.join(
            self.image_dir,
            image_id
        )

        image = Image.open(image_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, label
````

part 3

````python
from sklearn.model_selection import train_test_split
train_df, val_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))
````

````python
train_dataset = AgeDataset(
    train_df,
    image_dir,
    transform
)

val_dataset = AgeDataset(
    val_df,
    image_dir,
    transform
)
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False
)

print("created successfully.")
````

part D

````python
class AgeCNN(nn.Module):

    def __init__(self):
        super(AgeCNN, self).__init__()

        self.block1 = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),

            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),

            nn.MaxPool2d(2)
        )


        self.block2 = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),

            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),

            nn.MaxPool2d(2)
        )


        self.block3 = nn.Sequential(
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),

            nn.Conv2d(128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),

            nn.MaxPool2d(2)
        )


        self.block4 = nn.Sequential(
            nn.Conv2d(128, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),

            nn.Conv2d(256, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),

            nn.MaxPool2d(2)
        )
        self.fc = nn.Sequential(
            nn.Flatten(),

            nn.Linear(256 * 4 * 4, 512),
            nn.ReLU(),

            nn.Dropout(0.5),

            nn.Linear(512, 3)
        )

    def forward(self, x):

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)

        x = self.fc(x)

        return x
````

part E

````python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
model = AgeCNN().to(device)
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print(model)
print("Device:", device)
````

````python
epochs = 5

train_losses = []
val_losses = []

for epoch in range(epochs):

    model.train()

    train_loss = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        train_loss += loss.item()

    train_loss = train_loss / len(train_loader)

    model.eval()

    val_loss = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            loss = criterion(outputs, labels)

            val_loss += loss.item()

    val_loss = val_loss / len(val_loader)

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    print(
        f"Epoch {epoch+1}/{epochs} "
        f"Train Loss: {train_loss:.4f} "
        f"Val Loss: {val_loss:.4f}"
    )
````

````python
plt.plot(train_losses, label="Training Loss")
plt.plot(val_losses, label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")

plt.legend()
plt.show()
````

````python
model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in val_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total

print(f"Validation Accuracy: {accuracy:.2f}%")
````
